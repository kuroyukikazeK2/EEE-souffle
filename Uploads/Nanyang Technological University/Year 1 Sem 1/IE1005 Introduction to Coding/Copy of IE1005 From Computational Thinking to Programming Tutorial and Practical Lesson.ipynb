```python
/*10% of grade assignment 1*/
/* Distance travelled by a particle falling under gravity. Initial velocity = 0 */
#include <stdio.h>

int main(void)
{
double initial_velocity, max_time, interval, time = 0, ACCEL = 9.8, distance;
printf("Please only type in integers and floats within 2.2E-308 to 1.8E+308.\n" );
printf("Please input the initial velocity.\n" );
scanf("%lf",&initial_velocity);
printf("Please input the maximum time. The time must be a positive number.\n" );
scanf("%lf",&max_time);
printf("Please input the interval.\n" );
scanf("%lf",&interval);

while (time>max_time)
{
    printf("Please input the maximum time again. The time must be a positive number.\n" );
    scanf("%lf",&max_time);
}

while (interval>max_time)
{
    printf("Please input the interval again. The maximum time must be more than the interval.\n" );
    scanf("%lf",&interval);
}

printf("Time Elapsed Distance Travelled\n" );
printf("--------------------------------\n" );
while (time <= max_time)
{
distance = initial_velocity*time+0.5*ACCEL*time*time;
printf("%8.2f %8.2f\n", time, distance);
time = time + interval;
}
return 0;
}
```

```python
#include <stdio.h>
int second(const int arr[], int size);

int main(void)
{
  int a[6] = {1, 2, 3, 14, 15, 5};

  printf("\nARRAY\n" );
  printf("%d",second(a,6));
  return 0;
}

int second(const int arr[], int size)
{
    int num1,num2,i;
    if (arr[0]>arr[1])
    {
        num1=arr[0];num2=arr[1];
    }
    else
    {
        num1=arr[1];num2=arr[0];
    }
    for (i=2; i <size; i++)
    {
        if (arr[i]>num1)
        {
            num2=num1;
            num1=arr[i];
        }
        else if (arr[i]>num2)
        {
            num2 = arr[i];
        }
    }
    return num2;
}
```

```python
#include <stdio.h>
void add_matrices(int a[3][3],int b[3][3],int c[3][3]);
void print_matrix(int a[3][3]);
int main(void)
{
  int p[3][3] = {{1, 3, -4}, {1,1,-2},{-1,-2,5}};
  int q[3][3] = {{8,3,0}, {3,10,2},{0,2,6}};
  int r[3][3];
  add_matrices(p,q,r);
  printf("\np:\n");
  print_matrix(p);
  printf("\nq:\n");
  print_matrix(q);
  printf("\nr:\n");
  print_matrix(r);

  return 0;
}

void add_matrices(int a[3][3],int b[3][3],int c[3][3])
{
    int i,j;
    for (i=0; i<3; i++)
    {
        for (j=0; j<3; j++)
        {
            c[i][j]=a[i][j]+b[i][j];
        }
    }
}
void print_matrix(int a[3][3])
{
    int i,j;
    for (i=0; i<3; i++)
    {
        for (j=0; j<3; j++)
        {
            printf("%d\t",a[i][j]);
        }
        printf("\n");
    }
}
```

```python
#include <stdio.h>
void mult_matrices(int a[3][3],int b[3][3],int c[3][3]);
void print_matrix(int a[3][3]);
int main(void)
{
  int p[3][3] = {{1, 3, -4}, {1,1,-2},{-1,-2,5}};
  int q[3][3] = {{8,3,0}, {3,10,2},{0,2,6}};
  int r[3][3];
  mult_matrices(p,q,r);
  printf("\np:\n");
  print_matrix(p);
  printf("\nq:\n");
  print_matrix(q);
  printf("\nr:\n");
  print_matrix(r);

  return 0;
}

void mult_matrices(int a[3][3],int b[3][3],int c[3][3])
{
    int i,j,k;
    for (i=0; i<3; i++)
    {
        for (j=0; j<3; j++)
        {
            c[i][j]=a[i][0]*b[0][j];
            for (k=1; k<3; k++)
            {
                c[i][j]+=a[i][k]*b[k][j];
            }
        }
    }
}
void print_matrix(int a[3][3])
{
    int i,j;
    for (i=0; i<3; i++)
    {
        for (j=0; j<3; j++)
        {
            printf("%d\t",a[i][j]);
        }
        printf("\n");
    }
}
```
